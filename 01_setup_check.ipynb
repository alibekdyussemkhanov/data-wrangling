Setup check: load the tips dataset and plot the average bill by day.

```python
import pandas as pd
import matplotlib.pyplot as plt

tips = pd.read_csv("data/raw/tips.csv")
tips.head()
```

```python
day_order = ["Thur", "Fri", "Sat", "Sun"]
avg_bill = tips.groupby("day")["total_bill"].mean().reindex(day_order)
avg_bill.plot(kind="bar")
plt.title("Average bill by day")
plt.ylabel("Average total bill, USD")
plt.show()
```
